```python
import numpy as np

#1
a1 = np.empty((2,2))
print("1. Empty 2x2:\n", a1)

#2
a2 = np.ones((4,2))
print("\n2. Ones 4x2:\n", a2)

#3
a3 = np.full((3,3), 7)
print("\n3. Full matrix (fill=7):\n", a3)

#4
arr = np.array([[1,2,3],[4,5,6]])
a4 = np.zeros_like(arr)
print("\n4. Zeros_like(arr):\n", a4)

#5
a5 = np.ones_like(arr)
print("\n5. Ones_like(arr):\n", a5)

#6
new_list = [1,2,3,4]
a6 = np.array(new_list)
print("\n6. List to numpy array:\n", a6)
```

```python
#1
p1 = np.arange(10, 50)
print("1. 10 to 49:\n", p1)

#2
p2 = np.arange(9).reshape(3,3)
print("\n2. 3x3 0-8:\n", p2)

#3
p3 = np.eye(3)
print("\n3. Identity:\n", p3)

#4
p4 = np.random.random(30)
print("\n4. Mean of random 30:\n", p4.mean())

#5
p5 = np.random.random((10,10))
print("\n5. Min, Max:", p5.min(), p5.max())

#6
p6 = np.zeros(10)
p6[4] = 1
print("\n6. Zero array replace 5th with 1:\n", p6)

#7
arr = np.array([1,2,0,0,4,0])
p7 = arr[::-1]
print("\n7. Reverse array:\n", p7)

#8
p8 = np.ones((5,5))
p8[1:-1,1:-1] = 0
print("\n8. Border 1 inside 0:\n", p8)

#9
p9 = np.zeros((8,8))
p9[1::2, ::2] = 1
p9[::2, 1::2] = 1
print("\n9. Checkerboard:\n", p9)
```

```python
x = np.array([[1,2],[3,5]])
y = np.array([[5,6],[7,8]])
v = np.array([9,10])
w = np.array([11,12])

#1
print("1. x+y:\n", x+y)

#2
print("\n2. x-y:\n", x-y)

#3
print("\n3. x*3:\n", x*3)

#4
print("\n4. Square of x:\n", x**2)

#5
print("\n5. Dot products:")
print("v·w =", np.dot(v,w))
print("x·v =", x.dot(v))
print("x·y =\n", x.dot(y))

#6
print("\n6. Concatenate x & y (rows):\n", np.vstack((x,y)))
print("\nConcatenate v & w (columns):\n", np.column_stack((v,w)))

#7
try:
    print(np.concatenate((x,v)))
except Exception as e:
    print("\n7. Error Reason:", e)
```

```python
import numpy as np

# Given matrices
A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])

# 1. Prove A * A^-1 = I
print("A * A^-1 =\n", A.dot(np.linalg.inv(A)))

# 2. Prove AB != BA
print("\nAB =\n", A.dot(B))
print("\nBA =\n", B.dot(A))

# 3. Prove (AB)^T = B^T A^T
AB = A.dot(B)
print("\n(AB)^T =\n", AB.T)
print("\nB^T A^T =\n", B.T.dot(A.T))

# Linear equations:
# 2x − 3y + z = −1
# x − y + 2z = −3
# 3x + y − z = 9

C = np.array([[2,-3,1],
              [1,-1,2],
              [3,1,-1]])

d = np.array([-1,-3,9])

# Solve using inverse
solution = np.linalg.inv(C).dot(d)
print("\nSolution of the system =\n", solution)
```
