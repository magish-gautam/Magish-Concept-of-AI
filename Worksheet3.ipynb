```python
# SECTION 2 — EDA Practice Exercise 1

df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
#1)

df['fare_per_year'] = df['Fare'] / df['Age']
```

```python
#2)
high_fare_age = df[df['fare_per_year'] > 5]
```

```python
#3)
high_fare_age_srt = high_fare_age.sort_values(by='fare_per_year', ascending=False)

```

```python
#4)
result = high_fare_age_srt[['Name','fare_per_year']]
result.head()
```

```python
#Q2: Which adult male paid the highest fare relative to class?
# 1)
df['fare_per_class'] = df['Fare'] / df['Pclass']
```

```python
# 2)

adult_males = df[(df['Sex'] == 'male') & (df['Age'] >= 18)]
```

```python
# Q3)
adult_males_srt = adult_males.sort_values(by='fare_per_class', ascending=False)
```

```python
#Q4)

result2 = adult_males_srt[['Name','Age','fare_per_class']]
result2.head()
```

```python
# SECTION 3 — Group-By Analysis Practice

# Q1)
total_fare = df['Fare'].sum()
total_fare

```

```python
# Q2)

fare_by_class = df.groupby('Pclass')['Fare'].sum()
fare_by_class
```

```python
#Q3)
fare_percentage = (fare_by_class / total_fare) * 100
fare_percentage
```

```python
#Q2: Percentage of passengers by age group
#1)

df['age_group'] = pd.cut(
    df['Age'],
    bins=[0, 18, 65, 100],
    labels=['child','adult','senior'],
    include_lowest=True
)
```

```python
#Q2)
total_passengers = len(df)
```

```python
# Q3
age_group_counts = df['age_group'].value_counts()
age_group_counts
```

```python
# Q4)
age_group_percentage = (age_group_counts / total_passengers) * 100
age_group_percentage
```
